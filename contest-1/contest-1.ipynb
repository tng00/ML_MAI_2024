```python
import pandas as pd
```

```python
data = pd.read_csv('/kaggle/input/contest-1-datasets/train.csv')
```

```python
data.info()
```

```python
data.head()
```

Удалим строки с пропусками в датафрейме.

```python
data = data.dropna()
```

Кодируем категориальные переменные с помощью LabelEncoder, который преобразует их в целые числовые значения.

```python
from sklearn.preprocessing import LabelEncoder



le = LabelEncoder()



for column in data.drop(['ApplicationDate'], axis=1).columns:

    if data[column].dtype == 'object':

        data[column] = le.fit_transform(data[column])
```

Визуализируем выбросы в столбце RiskScore.

```python
import matplotlib.pyplot as plt

import seaborn as sns



sns.boxplot(y='RiskScore', data=data)

plt.title('Boxplot of RiskScore')

plt.show()
```

Удалим выбросы в столбце 'RiskScore' с помощью метода IQR.

```python
Q1 = data['RiskScore'].quantile(0.25)

Q3 = data['RiskScore'].quantile(0.75)

IQR = Q3 - Q1

data = data[~((data['RiskScore'] < (Q1 - 1.5 * IQR)) | (data['RiskScore'] > (Q3 + 1.5 * IQR)))]
```

Выведем список пар столбцов с высокой корреляцией из матрицы корреляции.

```python
corr_matrix = data.drop(['ApplicationDate'], axis=1).corr()

high_corr = corr_matrix[(corr_matrix >= 0.7) & (corr_matrix != 1.)].stack()

high_corr
```

Удалим некоторые столбцы, чтобы избежать мультиколлинеарности.

```python
to_drop = ['ApplicationDate', 'RiskScore', 'AnnualIncome', 'Experience', 'InterestRate', 'CreditScore', 'TotalAssets', 'LoanAmount']

X = data.drop(to_drop, axis=1)

y = data['RiskScore']
```

Обучим модель линейной регрессии.

```python
from sklearn.linear_model import LinearRegression



model = LinearRegression()

model.fit(X, y)
```

Аналогичным образом подготовим тестовые данные.

```python
test_data = pd.read_csv('/kaggle/input/contest-1-datasets/test.csv')
```

```python
for column in test_data.drop(['ApplicationDate'], axis=1).columns:

    if test_data[column].dtype == 'object':

        test_data[column] = le.fit_transform(test_data[column])
```

```python
test_data.info()
```

```python
to_drop = ['ApplicationDate', 'ID', 'AnnualIncome', 'Experience', 'InterestRate', 'CreditScore', 'TotalAssets', 'LoanAmount']
```

Предскажем значения целевой переменной для тестовой выборки данных с помощью обученной модели линейной регрессии.

```python
X_test = test_data.drop(to_drop, axis=1)

y_test = model.predict(X_test)

y_test
```

Сохраним предсказанные значения целевой переменной в файл CSV.

```python
ids = range(5000)
df = pd.DataFrame({'ID': ids, 'RiskScore': y_test})
df.to_csv('submission.csv', index=False)
```
